A tensor is a mathematical object and data structure representing an 
-dimensional array of numbers, generalizing scalars (0D), vectors (1D), and matrices (2D) 
to higher dimensions. Used extensively in physics and machine learning (e.g., TensorFlow), 
tensors organize complex data, with their rank defining the number of dimensions. 

Data Representation: Tensors hold data in 
-dimensions, making them suitable for representing complex, multi-directional physical properties (like stress or conductivity) or data in neural networks.

<b>Zeros and ones<b>

```python
shape = (2, 3,)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

```python

rand_tensor = torch.rand(size = (2, 3))
ones_tensor = torch.ones(size = (2, 3))
zeros_tensor = torch.zeros(size = (2, 3))

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

```python
ones_tensor.dtype
```
