```python
# PICKLE MODULE

import pickle 
data = {'name': 'Matt Murdock', 'age': 22}

file = open('data.pkl', 'wb')
pickle.dump(data, open('data.pkl', 'wb'))

file2 = open('data.pkl', 'rb')
load_data = pickle.load(file2)
print(load_data)
```

```python
# PICKLING WITHOUT A FILE

import pickle

Leo = {'key' : 'Leo', 'name' : 'Leo Johnson', 
'age' : 21, 'pay' : 40000}
Harry = {'key' : 'Harry', 'name' : 'Harry Jenner',
'age' : 50, 'pay' : 50000}

db = {}
db['Leo'] = Leo
db['Harry'] = Harry

b = pickle.dumps(db)   
myEntry = pickle.loads(b)
print(myEntry)
```

```python
# OPERATIONS ON CSV FILES

import csv
file = open('data.csv', 'w', newline='')
writer = csv.writer(file)
writer.writerow(['Name', 'Age'])
writer.writerow(['Luffy', '18'])
writer.writerow(['Zoro', '19'])
writer.writerow(['Nami', '18'])
writer.writerow(['Ussop', '18'])
writer.writerow(['Sanji', '21'])
```

```python
file = open('data.csv', 'r')
reader = csv.reader(file)
for row in reader:
    print(row)
# reading dictionaries
```

```python
# reading dictionaries

with open('data.csv', mode='r', newline='', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(dict(row))
```
