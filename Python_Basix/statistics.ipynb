```python
import numpy as np
```

```python
arr = [3,7,5,8,2]
mean = np.mean(arr)
print(mean)
```

```python
import scipy.stats as stats
arr = [2, 3, 3, 5, 7]
mode = stats.mode(arr)
print(mode)
```

```python
median = np.median(arr)
print(median)
```

```python
# Range

arr = [3,6,4,9,7]

Max = max(arr)
Min = min(arr)
range = Max - Min
print(f"max : {Max}, min :{Min} and range : {range}")
```

```python
# Varience

import statistics

arr = [1, 2, 3, 4, 5]
print("var :", (statistics.variance(arr)))
```

```python
# standard deviation
''' Standard deviation quantifies how much individual data points deviate from the mean, A high 
standard deviation means greater dispersion while a low standard deviation indicates that 
data points are close to mean'''
import statistics
arr = [1, 2, 3, 4, 5]
print("Std = ", (statistics.stdev(arr)))
```
