```python
import numpy as py
import pandas as pd
import matplotlib.pyplot as plt
```

```python
cols = ["fLength","fWidth","fSize","fConcl","fAysm","fM3Long","fM3Trans","fAlpha","fDist","class"]
df = pd.read_csv("magic04.data", names = cols)
df.head()                #gives top 5 rows
```

```python
df['class'] = (df['class'] == "g").astype(int)
```

```python
df.head()
```

```python
for label in cols[:-1]:
    plt.hist(df[df["class"]==1][label], color = 'green', label = 'gamma', alpha = 0.6, density = True)
    plt.hist(df[df["class"]==0][label], color = 'red', label = 'hadron', alpha = 0.6, density = True)
    plt.title(label)
    plt.ylabel("probability")
    plt.xlabel(label)
    plt.legend()
    plt.show()
```
