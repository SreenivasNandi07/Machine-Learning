```python
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.add(a, b)
print(res)
```

```python
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.subtract(a, b)
print(res)
```

```python
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.multiply(a, b)
print(res)
```

```python
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.divide(a, b)
print(res)
```

```python
#power
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.power(a, b)
print(res)
```

```python
#modulus
a = np.array([5, 72, 13, 100])
b = np.array([2, 5, 10, 30])
res = np.mod(a, b)
print(res)
```

```python
#Broadcasting a Scalar to a 1D Array
arr = np.array([1, 2, 3])
res = arr + 1  
print(res)

#Broadcasting a 1D Array to a 2D Array
a = np.array([2, 4, 6])
b = np.array([[1, 3, 5], [7, 9, 11]])
res = a + b
print(res)

#Broadcasting in Conditional Operations
a = np.array([12, 24, 35, 18, 60, 72])
b = np.array(["Adult", "Minor"])
res = np.where(a > 18, b[0], b[1])
print(res)
```

```python
# Adjusting Temperature Data Across Multiple Locations
temp = np.array([ [30, 32, 34, 33, 31],
                  [25, 27, 29, 28, 26],
                  [20, 22, 24, 23, 21] ])

corr = np.array([1.5, -0.5, 2.0])
res = temp + corr[:, None]
print(res)
#Normalizing Image Data
img = np.array([ [100, 120, 130],
                 [90, 110, 140],
                 [80, 100, 120] ])

m = img.mean(axis=0)
s = img.std(axis=0)
res = (img - m) / s
print(res)
#Centering Data in Machine Learning
data = np.array([ [10, 20],
                  [15, 25],
                  [20, 30] ])

m = data.mean(axis=0)
res = data - m
print(res)
```
