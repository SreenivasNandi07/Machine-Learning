```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({
    'Hours': [2, 4, 6, 8, 7, 3],
    'Attendance': [60, 70, 75, 85, 78, 65],
    'Result': ['Fail', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
})

X = df[['Hours','Attendance']]

scaler = MinMaxScaler()
Xscale = scaler.fit_transform(X)
Xscale = pd.DataFrame(Xscale, columns = ['Hours','Attendance'])
print(Xscale, df['Result'])
```

```python
plt.scatter(X)
```

```python
new = [[1, 1]]
newscale = scaler.transform(new)
```

```python
knn = KNeighborsClassifier(n_neighbors = 3)
knn.fit(Xscale, df['Result'])

pred = knn.predict(newscale)
print(pred)
```
