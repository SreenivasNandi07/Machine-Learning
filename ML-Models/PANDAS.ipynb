<b> Binary operations on Pandas Series <b>

```python
ser1 = pd.Series([1, 2, 3], index=['A', 'B', 'C'])
ser2 = pd.Series([4, 5, 6], index=['A', 'B', 'C'])

df_sum = ser1.add(ser2)
print(df_sum)

df_sum = ser1.mul(ser2)
print(df_sum)

df_sum = ser1.div(ser2)
print(df_sum)
# other operations like sub(), sum(){Returns the sum of the values for the requested axis}, prod()
#mean(), pow(), abs(),cov()
```

<b>Conversion operations on Series<b>

```python
ser = pd.Series([1, 2, 3, 4])
ser = ser.astype(float)
print(ser)
```

<b> Pandas CRUD operations <b>

```python
# creating
data = {
    "Name": ["Ansh", "Sahil", "Ram"],
    "Age": [21, 20, 41],
    "City": ["Moradabad", "New Delhi", "Chennai"]
}
df = pd.DataFrame(data)
print(df)

# creating from lists
names = ["Akshit", "Uday", "Sam"]
ages = [25, 30, 35]
cities = ["Gurugram", "New Delhi", "Chicago"]

df = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "City": cities
})

print(df)
```

```python
#reading a dataframe

print(df.head(3))
print('')
print(df.tail(3))
print('')
print(df.columns)
print('')
print(df.dtypes)
print('')
print(df.describe())
print('')
```

```python
print(df["Name"])
print('')
print(df[["Name", "City"]])
print('')
print(df["City"].unique())
print('')
print(df[df["Age"] > 30])
print('')
print(df[(df["Age"] > 30) & (df["City"] == "SF")])

print(df[(df["Age"] > 30) | (df["City"] == "LA")])
print('')
print(df.iloc[0])
```

<b>Modifying data in pandas<b>

```python
data = {'Name': ['Eve', 'Jack', 'Charlie', 'Henry', 'John'],
        'Age': [25, 30, 35, 40, 45],
        'City': ['NY', 'LA', 'SF', 'Houston', 'Seattle']}
df = pd.DataFrame(data)
df
```

```python
# Updating a single value
df.loc[df['Name'] == 'Jack', 'Age'] = 42

print(df)
```

```python
#updating entire column
df['City'] = ['Boston', 'Chicago', 'LA', 'Austin', 'Miami']
df
```

```python
# updating based on condition
df.loc[df['City'] == 'LA', 'Age'] = 31
df
```

<b>Deleting data in pandas<b>

```python
data = {'Name': ['Eve', 'Jack', 'Charlie', 'Henry', 'John'],
        'Age': [25, 30, 35, 40, 45],
        'City': ['NY', 'LA', 'SF', 'Houston', 'Seattle']}
df = pd.DataFrame(data)

print(df)
```

```python
#Deleting a column
df = df.drop('City', axis=1)
print(df)
```

```python
#Deleting a Row
df = df.drop(2, axis=0) 
print(df)
```

```python
#Deleting based on condition
df = df[df['Age'] != 35]
print(df)
```

```python
#Delete entire dataset
df = pd.DataFrame(data)

del df
```
