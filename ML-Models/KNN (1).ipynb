```python
import numpy as py
from collections import Counter
```

```python
distances = []
def knn_predict(traindata, trainlabels, testp, k):
    for i in range(len(traindata)):
        dist =  py.sqrt(py.sum((py.array(testp) - py.array(traindata[i]))**2))
        #euclidean_dis(testp,traindata[i])
        distances.append((dist, trainlabels[i]))
    distances.sort(key=lambda x: x[0])
    k_nearest_labels = [label for _, label in distances[:k]]
    return Counter(k_nearest_labels).most_common(1)[0][0]
```

```python
traindata = [[1, 2], [2, 3], [3, 4], [6, 7], [7, 8]]
trainlabels = ['A', 'A', 'A', 'B', 'B']
testp = [4, 5]
k = 3
```

```python
prediction = knn_predict(traindata, trainlabels, testp, k)
print(prediction)
```
